## Cross-File Comparison: Performance Files vs. employee_attrition.csv

> **Read-only analysis.** No merges, no drops. We are only checking whether join keys actually overlap.

The three performance-shaped files are:
- `Cleaned_HR_Data_Analysis.csv` -> `hr_clean`
- `Employee_Performance_Dataset.csv` -> `perf_ds`
- `employee_performance_pro.csv` -> `perf_pro`

We compare each against `employee_attrition.csv` (`attrition`) on:
1. Row counts
2. ID-overlap with `attrition['EmployeeNumber']` (set intersection - no name-similarity assumptions)
3. Engagement/satisfaction-relevant column richness

```python
# STEP A: Row-count comparison
print('=== ROW COUNT COMPARISON ===')
frames = {
    'employee_attrition':           attrition,
    'Cleaned_HR_Data_Analysis':     hr_clean,
    'Employee_Performance_Dataset': perf_ds,
    'employee_performance_pro':     perf_pro,
}
for name, df in frames.items():
    print(f'  {name:42s}  rows={df.shape[0]:>6,}  cols={df.shape[1]:>3}')
```

```python
# STEP B: ID-overlap check
# Each performance file is tested against attrition['EmployeeNumber'].
# We do NOT assume column names match; we test EVERY ID-like column
# in each performance file via set intersection.

# Determine the anchor set from attrition
if 'EmployeeNumber' in attrition.columns:
    anchor_col = 'EmployeeNumber'
else:
    id_candidates = [c for c in attrition.columns if 'id' in c.lower()]
    anchor_col = id_candidates[0] if id_candidates else None
    if anchor_col:
        print(f'WARNING: EmployeeNumber not found in attrition -- using "{anchor_col}" as anchor instead.')

overlap_results = {}

if anchor_col:
    anchor_vals = set(attrition[anchor_col].dropna().astype(str))
    anchor_n    = len(anchor_vals)
    print(f'Anchor: attrition["{anchor_col}"]  --> {anchor_n} unique values')
    print()

    perf_frames = {
        'Cleaned_HR_Data_Analysis':     hr_clean,
        'Employee_Performance_Dataset': perf_ds,
        'employee_performance_pro':     perf_pro,
    }

    print('=== ID OVERLAP CHECK (set intersection, no name-similarity assumptions) ===')

    for fname, df in perf_frames.items():
        print(f'\n--- {fname} ---')
        id_like = [c for c in df.columns if 'id' in c.lower()]

        if not id_like:
            print(f'  No ID-like columns found. Overlap cannot be computed.')
            overlap_results[fname] = []
            continue

        file_results = []
        for col in id_like:
            candidate_vals = set(df[col].dropna().astype(str))
            intersection   = anchor_vals & candidate_vals
            n_overlap      = len(intersection)
            pct_of_anchor  = (n_overlap / anchor_n       * 100) if anchor_n       else 0.0
            pct_of_cand    = (n_overlap / len(candidate_vals) * 100) if candidate_vals else 0.0
            file_results.append({
                'perf_col':               col,
                'n_unique_in_perf_file':  len(candidate_vals),
                'overlap_count':          n_overlap,
                'pct_of_attrition_anchor': round(pct_of_anchor, 2),
                'pct_of_perf_col':        round(pct_of_cand, 2),
            })
            print(f'  {col:38s}  n_unique={len(candidate_vals):>6}  '
                  f'overlap={n_overlap:>5}  '
                  f'(%attrition={pct_of_anchor:5.1f}%  %perf_col={pct_of_cand:5.1f}%)')
        overlap_results[fname] = file_results
else:
    print('Cannot compute ID overlap -- no usable anchor column found in attrition.')
```

```python
# STEP C: Engagement / satisfaction column richness
# Keywords drawn from HR literature
ENGAGEMENT_KEYWORDS = [
    'satisf', 'engage', 'wellb', 'morale', 'motivat',
    'burnout', 'commit', 'nps', 'survey', 'sentiment',
    'happi', 'loyal', 'retention', 'turnover', 'balance',
    'environment', 'relation', 'climate',
]

def count_engagement_cols(df):
    matched = [c for c in df.columns
               if any(kw in c.lower() for kw in ENGAGEMENT_KEYWORDS)]
    return matched

print('=== ENGAGEMENT / SATISFACTION COLUMN RICHNESS ===')
for fname, df in {
    'Cleaned_HR_Data_Analysis':     hr_clean,
    'Employee_Performance_Dataset': perf_ds,
    'employee_performance_pro':     perf_pro,
}.items():
    matched = count_engagement_cols(df)
    print(f'\n  {fname}  -->  {len(matched)} engagement-related column(s):')
    if matched:
        for c in matched:
            print(f'    * {c}')
    else:
        print('    (none matched)')
```

## Final Summary & Recommendation

> **Read-only analysis complete.** The table and recommendation below are based solely on evidence gathered in this notebook. No merges or cleaning have been performed.

### Dataset Summary Table

*(Programmatic table is generated in the next cell from live data.  
The static table below acts as a reference skeleton.)*

| # | Dataset | Rows | Cols | Likely Primary Key | Top Missing Col (% miss) | Purpose |
|---|---------|------|------|-------------------|--------------------------|---------|
| 1 | `employee_attrition.csv` | — | — | `EmployeeNumber` | — | Attrition target + HR features (**anchor**) |
| 2 | `occupation_data.csv` | — | — | O*NET / SOC code | — | External occupation-level skill profiles |
| 3 | `essential_skills.csv` | — | — | Skill / O*NET code | — | Task/skill taxonomy (likely occupational) |
| 4 | `software_skills.csv` | — | — | Software / O*NET code | — | Software tool taxonomy (likely occupational) |
| 5 | `Cleaned_HR_Data_Analysis.csv` | — | — | ID col → see step 6 | — | Performance/HR analysis |
| 6 | `Employee_Performance_Dataset.csv` | — | — | ID col → see step 6 | — | Employee performance records |
| 7 | `employee_performance_pro.csv` | — | — | ID col → see step 6 | — | Performance extended (pro) |

```python
# PROGRAMMATIC SUMMARY TABLE — live values from loaded dataframes

def top_missing_col(df):
    pct = (df.isnull().sum() / len(df) * 100).sort_values(ascending=False)
    if pct.iloc[0] == 0:
        return 'none', 0.0
    return pct.index[0], round(pct.iloc[0], 1)

def likely_pk(df):
    id_like = [c for c in df.columns if 'id' in c.lower()]
    if not id_like:
        return 'none detected'
    return max(id_like, key=lambda c: df[c].nunique())

summary_rows = []
for label, df in [
    ('employee_attrition',           attrition),
    ('occupation_data',              occ),
    ('essential_skills',             ess),
    ('software_skills',              sw),
    ('Cleaned_HR_Data_Analysis',     hr_clean),
    ('Employee_Performance_Dataset', perf_ds),
    ('employee_performance_pro',     perf_pro),
]:
    mc, mp = top_missing_col(df)
    summary_rows.append({
        'Dataset':         label,
        'Rows':            df.shape[0],
        'Cols':            df.shape[1],
        'Likely PK':       likely_pk(df),
        'Top Missing Col': mc,
        'Top Missing %':   mp,
        'Engagement Cols': len(count_engagement_cols(df)),
    })

summary_df = pd.DataFrame(summary_rows).set_index('Dataset')
display(summary_df)
```

```python
# EVIDENCE-DRIVEN RECOMMENDATION
# Scoring: overlap with attrition[EmployeeNumber] weighted 60%,
#          engagement/satisfaction column count weighted 40%.
# No merge is performed here.

print('=' * 72)
print('RECOMMENDATION: Which performance file(s) to carry into data/processed/')
print('=' * 72)

_perf_map = {
    'Cleaned_HR_Data_Analysis':     hr_clean,
    'Employee_Performance_Dataset': perf_ds,
    'employee_performance_pro':     perf_pro,
}

scores = {}
for fname, df in _perf_map.items():
    eng_cols     = len(count_engagement_cols(df))
    results      = overlap_results.get(fname, [])
    best_overlap = max((r['pct_of_attrition_anchor'] for r in results), default=0.0)
    score        = best_overlap * 0.6 + eng_cols * 0.4
    scores[fname] = {
        'best_overlap_pct': best_overlap,
        'engagement_cols':  eng_cols,
        'composite_score':  round(score, 2),
    }
    print(f'  {fname}')
    print(f'    Best ID overlap with attrition[EmployeeNumber] : {best_overlap:.1f}%')
    print(f'    Engagement / satisfaction columns              : {eng_cols}')
    print(f'    Composite score  (overlap*0.6 + eng_cols*0.4) : {score:.2f}')
    print()

winner = max(scores, key=lambda k: scores[k]['composite_score'])

print('-' * 72)
print(f'TOP RECOMMENDATION --> {winner}')
print()
print('Rationale:')
print('  1. ID OVERLAP: A non-zero overlap % with attrition[EmployeeNumber] confirms')
print('     the files share actual employee records, making a future join sound.')
print('     Files with 0% overlap are EXTERNAL reference tables, not row-level')
print('     join candidates, and must NOT be merged on ID in Step 3.')
print()
print('  2. ENGAGEMENT RICHNESS: More engagement/satisfaction columns means richer')
print('     feature signal for modelling attrition, retention risk, and burnout.')
print()
print('  3. If ALL three files score 0% overlap (e.g. fully synthetic data),')
print('     select by highest engagement col count + lowest % missing, and flag')
print('     explicitly in the Step 2 data dictionary that no verified join key')
print('     exists -- the feature engineering step must create a synthetic bridge.')
print('=' * 72)
```
